```python
import numpy as np
```

```python
#Premier exemple d'implémentation de la Descente des Coordonnées

#Itéré initial
x = np.array([0.5,-1])

for k in range(10):
    #Mettre à jour x(0)
    x[0] =(6*x[1]+4)/10

    #Mettre à jour x(1)
    x[1] =(6*x[0]-3)/10

    print(f"A l'itéré {k}, on a x = {x}")
```

```python
def CoordinateDescentForQuadratic(Q,c,p,x,maxiter):
  n = len(x)

  for k in range(maxiter):
    #A chaque itéré k, on met toutes les variables à jour une fois de façon cyclique
    #Un itéré coute O(n^2)
    for i in range(n):
      somme = 0 #le calcul de somme coute n
      for j in range(n):
        somme += Q[i,j]*x[j]
      x[i] = x[i] - (somme - c[i])/(Q[i,i])
    print(f"A l'itéré {k}, on a x = {x}")
  return x

  Q = np.array([[10,-6],[-6,10]])
  c = np.array([4,3])
  p = 0
  x0 = np.array([-0.5,-1])
  maxiter = 10
  x = CoordinateDescentForQuadratic(Q,c,p,x0,maxiter)
```
